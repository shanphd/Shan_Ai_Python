### Python Positional vs Keyword Arguments


```python
from IPython.display import Image,display
display(Image("42.2.png",width=400))
```

```python
def clean_name  (fn,ln,country):
    f = fn.strip().lower()
    l = ln.strip().lower()
    c = f + " " + l + " " + country
    print(f"clean_name: {c}")
```

```python
#positional 

clean_name("John ","DON ", "IND ")

#keyword 

clean_name (country = "IND", ln = "DON ", fn = "SHAN ")
```

```python
# Default 

def clean_name  (fn,ln,country='NA'):
    f = fn.strip().lower()
    l = ln.strip().lower()
    c = f + " " + l + " " + country
    print(f"clean_name: {c}")
```

```python
clean_name ("SHAN ", "JOE ")
```
