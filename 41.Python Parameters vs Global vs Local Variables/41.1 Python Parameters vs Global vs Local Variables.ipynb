```python
# Python Global and local variable 

case_rule = 'lower'

def clean_text(name):
    print(f"#"*10)
    clean = name.strip()
    if case_rule == 'lower':
        clean = clean.lower()
    print(f"The Rule is = {case_rule}")
    print(f"cleaned name : {clean}")
    print(f"#"*10)
        
```

```python
clean_text('Maria  ')
```
