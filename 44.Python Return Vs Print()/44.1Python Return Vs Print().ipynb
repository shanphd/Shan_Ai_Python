```python
# Return Function

def clean_name(name):
    lo_name = name.strip().lower()
    up_name = name.strip().upper()
    return lo_name,up_name

lo_name,up_name = clean_name(" SHaN ")
print(lo_name)
print(up_name)
```

```python
def sfk_name(sysid):
    if not sysid:
        return None
    else:
         lo_name = sysid.strip().lower()
         up_name = sysid.strip().upper()
         return lo_name,up_name
     
name = sfk_name("ITP")
print(name)
            
```
