```python
#* args

from math import prod
def mul_shan(*args):
    print(prod(args))
    
mul_shan(3,3)
```

```python
def div_shan(*args):
    result = args[0]
    for x in args[1:]:
        result /= x
    print(result)
```

```python
div_shan(100,5,2)
```

```python
def sfk_sysid(**kwars):
    print(kwars)
    
sfk_sysid(name='sfk',owner='Arjun',email_d='sfk@dtc.com',ph_no='866786890')
```
